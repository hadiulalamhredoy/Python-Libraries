```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv('dataset_lab04.csv')

# Display first 5 rows
print(df.head())
```

```python
# 1. How many rows and columns does this dataframe have?

row, col = df.shape

print("Number of rows =", row)
print("Number of columns =", col)
```

```python
# 2. Describe (Numerical Summary) the Time and Amount Columns
# Statistics: Count, Mean, Standard Deviation, Minimum, Maximum, 25th Percentile (Q1), 50th Percentile (Median / Q2), 75th Percentile (Q3)

print("Summary of Time Column:")
print(df["Time"].describe())

print("\nSummary of Amount Column::")
print(df["Amount"].describe())
```

```python
# 3. Compute Mean, Median, Standard Deviation, and Variance

print("Time Column:")
print("Mean = ", df["Time"].mean())
print("Median = ", df["Time"].median())
print("Standard Deviation = ", df["Time"].std())
print("Variance = ", df["Time"].var())

print("\nAmount Column:")
print("Mean = ", df["Amount"].mean())
print("Median = ", df["Amount"].median())
print("Standard Deviation = ", df["Amount"].std())
print("Variance = ", df["Amount"].var())

# For All Numerical Columns
stats = pd.DataFrame({
    "Mean": df.mean(numeric_only=True),
    "Median": df.median(numeric_only=True),
    "Std Dev": df.std(numeric_only=True),
    "Variance": df.var(numeric_only=True)
})

print(stats)
```

```python
# 4. Compute Mean Using Your Own Function and Compare with Pandas

# User-defined mean function
def my_mean(column):
    return sum(column) / len(column)

my_mean_amount = my_mean(df["Amount"])

pandas_mean_amount = df["Amount"].mean()

print("My Mean     =", my_mean_amount)
print("Pandas Mean =", pandas_mean_amount)

# Compare Difference
difference = abs(my_mean_amount - pandas_mean_amount)
print("Difference =", difference)
```

```python
# 5. Histogram of Time and Amount

# Histogram of Time
# plt.figure(figsize=(10, 5))
# plt.hist(df["Time"], bins=20)
# plt.title("Histogram of Time")
# plt.xlabel("Time")
# plt.ylabel("Frequency")
# plt.show()

# Histogram of Amount
plt.figure(figsize=(10, 5))
plt.hist(df["Amount"], bins=20)
plt.title("Histogram of Amount")
plt.xlabel("Amount")
plt.ylabel("Frequency")
plt.show()
```

```python
# 6. Find the percentage of rows with class value = 0 (Non-Fraudulent) and class value = 1 (Fraudulent)

class_counts = df["Class"].value_counts()

total_rows = len(df)

percentages = (class_counts / total_rows) * 100

print("Non-Fraudulent (Class=0):", percentages[0], "%")
print("Fraudulent (Class=1):", percentages[1], "%")
```

```python
# 7. Show the result you have got in 6 using a histogram.

percentages.plot(kind="bar")

plt.title("Fraud vs Non-Fraud Percentage")
plt.xlabel("Class")
plt.ylabel("Percentage")
plt.show()
```

```python
# 8. Show the histogram (data distribution) of a few other columns. Differentiate between left-skewed and right-skewed distributions.

# Example Columns
columns = ["V1", "V2", "V3", "V4"]

for col in columns:
    plt.figure(figsize=(6,4))
    plt.hist(df[col], bins=20)
    plt.title(f"Histogram of {col}")
    plt.xlabel(col)
    plt.ylabel("Frequency")
    plt.show()

# Calculate Skewness
# Skewness Value |	Distribution
#     > 0	     |  Right Skewed
#     < 0	     |  Left Skewed
#     ≈ 0	     |  Symmetric

for col in columns:
    skew_value = df[col].skew()
    
    print(col, "Skewness =", skew_value)

    if skew_value > 0:
        print("Right-Skewed Distribution\n")
    elif skew_value < 0:
        print("Left-Skewed Distribution\n")
    else:
        print("Symmetric Distribution\n")
```

```python
# 9. Find Positive Correlations Among Columns

corr_matrix = df.corr(numeric_only=True)

#print(corr_matrix)

# Extract Strong Positive Correlations
for col1 in corr_matrix.columns:
    for col2 in corr_matrix.columns:

        if col1 != col2:

            corr = corr_matrix.loc[col1, col2]

            if corr > 0.35:
                print(col1, "-", col2, "Correlation =", round(corr, 3))
```

```python
# 10. Support Positive Correlation Using BoxPlot
# The strongest positive correlation is between: V27 & V28

sns.boxplot(y=df["V7"])

plt.title("BoxPlot of V7")
plt.show()

sns.boxplot(y=df["Amount"])

plt.title("BoxPlot of Amount")
plt.show()
```

```python
# 11. Support Positive Correlation Using Scatter Plot
# Interpretation: A positive upward trend indicates positive correlation.
plt.figure(figsize=(6,5))

plt.scatter(df["V7"], df["Amount"])

plt.xlabel("V7")
plt.ylabel("Amount")

plt.title("Scatter Plot of V7 vs Amount")

plt.show() 
```

```python
# 12. Find Negative Correlations Among Columns
for col1 in corr_matrix.columns:
    for col2 in corr_matrix.columns:

        if col1 != col2:

            corr = corr_matrix.loc[col1, col2]

            if corr < -0.5:
                print(col1, "-", col2, "Correlation =", round(corr,3))
```

```python
# 13. Support Negative Correlation Using BoxPlot
# The strongest negative correlation is between: V2 & Amount

sns.boxplot(y=df["V2"])

plt.title("BoxPlot of V2")
plt.show()

sns.boxplot(y=df["Amount"])

plt.title("BoxPlot of Amount")
plt.show()
```

```python
# 14. Support Negative Correlation Using Scatter Plot
# Interpretation: A downward trend indicates negative correlation..
plt.figure(figsize=(6,5))

plt.scatter(df["V2"], df["Amount"])

plt.xlabel("V2")
plt.ylabel("Amount")

plt.title("Scatter Plot of V2 vs Amount")

plt.show() 
```

```python
# Bonus: Correlation Heatmap
# This provides a much clearer view of positive and negative correlations.
# Interpretation:
# Dark Red → Strong Positive Correlation
# Dark Blue → Strong Negative Correlation
# White → No Correlation

plt.figure(figsize=(12,10))

sns.heatmap(
    df.corr(numeric_only=True),
    cmap="coolwarm",
    center=0
)

plt.title("Correlation Heatmap")
plt.show()
```
